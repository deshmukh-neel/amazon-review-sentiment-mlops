```python
MONGO_URI = "mongodb://localhost:27017"
DB_NAME = "videogames"
COLLECTION_NAME = "metadata"
```

```python
def get_collection(MONGO_URI: str, DB_NAME: str, COLLECTION_NAME: str):
    """
    Establishes connection to MongoDB and returns the collection.
    DO NOT MODIFY THIS FUNCTION.
    """
    client = MongoClient(MONGO_URI)
    db = client[DB_NAME]
    return db[COLLECTION_NAME]
```

```python
client = MongoClient(MONGO_URI)
db = client[DB_NAME]
collection = get_collection(MONGO_URI, DB_NAME, COLLECTION_NAME)
```

```python
pipeline = [
  { "$match": { "price": { "$ne": "null" } } },
  { "$bucket": {
      "groupBy": "$price",
      "boundaries": [0, 10, 25, 50, 100, 500],
      "default": "500+",
      "output": {
        "avg_rating": { "$avg": "$average_rating" },
        "total_products": { "$sum": 1 },
        "avg_price": { "$avg": "$price" },
        "total_ratings": { "$sum": "$rating_number" }
      }
  }},
  { "$out": "price_range_analysis" }
]
```

```python
collection.aggregate(pipeline)
```

```python
COLLECTION_NAME = "price_range_analysis"
collection = get_collection(MONGO_URI, DB_NAME, COLLECTION_NAME)
```
